```python

import random
import math

N = 8

# Calculate attacking queen pairs
def calculate_cost(board):
    cost = 0

    for i in range(N):
        for j in range(i + 1, N):
            if board[i] == board[j]:
                cost += 1
            elif abs(board[i] - board[j]) == abs(i - j):
                cost += 1

    return cost


# Generate a random neighboring state
def get_neighbor(board):
    next_board = board.copy()

    col = random.randint(0, N - 1)
    row = random.randint(0, N - 1)

    while row == next_board[col]:
        row = random.randint(0, N - 1)

    next_board[col] = row
    return next_board


# Display the chessboard
def print_board(board):
    for row in range(N):
        for col in range(N):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


# Simulated Annealing algorithm
def simulated_annealing():
    current = [random.randint(0, N - 1) for _ in range(N)]
    T = 100.0
    cooling_rate = 0.995

    print("Initial state:", current)
    print("Initial cost:", calculate_cost(current))

    while T > 0.001:
        current_cost = calculate_cost(current)

        if current_cost == 0:
            break

        next_state = get_neighbor(current)
        next_cost = calculate_cost(next_state)

        # Cost difference as in the PPT
        delta_E = current_cost - next_cost

        if delta_E > 0:
            current = next_state
        else:
            probability = math.exp(delta_E / T)

            if random.random() < probability:
                current = next_state

        T *= cooling_rate

    print("\nFinal state:", current)
    print("Final cost:", calculate_cost(current))

    if calculate_cost(current) == 0:
        print("\nSolution found!")
        print_board(current)
    else:
        print("\nNo solution found. Try running again.")


simulated_annealing()
```
