```python

# 8 Puzzle using Depth First Search (DFS)

def dfs(start, goal, max_depth=30):
    stack = [(start, [start])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state == goal:
            print("Goal state found!")
            print("Number of moves:", len(path) - 1)

            for s in path:
                print_state(s)
            return True

        if len(path) - 1 >= max_depth:
            continue

        if state in visited:
            continue

        visited.add(state)

        zero = state.index(0)
        row, col = divmod(zero, 3)

        # Possible blank movements: up, down, left, right
        moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

        for dr, dc in moves:
            nr, nc = row + dr, col + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                new_pos = nr * 3 + nc
                new_state = list(state)

                new_state[zero], new_state[new_pos] = (
                    new_state[new_pos], new_state[zero]
                )

                new_state = tuple(new_state)

                if new_state not in visited:
                    stack.append((new_state, path + [new_state]))

    print("Goal not found within the depth limit.")
    return False


def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


start = (1, 2, 3,
         5, 0, 6,
         4, 7, 8)
goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

dfs(start, goal)
```

```python
def get_neighbors(state):
    zero = state.index(0)
    row, col = divmod(zero, 3)
    neighbors = []

    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:
        nr, nc = row + dr, col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_pos = nr * 3 + nc
            board = list(state)
            board[zero], board[new_pos] = board[new_pos], board[zero]
            neighbors.append(tuple(board))

    return neighbors


def depth_limited_search(state, goal, limit, path, path_set):
    if state == goal:
        return path[:]

    if limit == 0:
        return None

    for neighbor in get_neighbors(state):
        if neighbor not in path_set:
            path.append(neighbor)
            path_set.add(neighbor)

            result = depth_limited_search(
                neighbor, goal, limit - 1, path, path_set
            )

            if result is not None:
                return result

            path.pop()
            path_set.remove(neighbor)

    return None


def ids(initial, goal):
    limit = 0

    while True:
        path = [initial]
        path_set = {initial}

        result = depth_limited_search(
            initial, goal, limit, path, path_set
        )

        if result is not None:
            return result

        limit += 1


def print_solution(path):
    if path is None:
        print("No solution found.")
        return

    print("Solution found in", len(path) - 1, "moves.")

    for step, state in enumerate(path):
        print(f"\nStep {step}:")
        for i in range(0, 9, 3):
            print(*state[i:i + 3])


initial = (1, 2, 3, 0, 4, 6, 7, 5, 8)
goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

solution = ids(initial, goal)
print_solution(solution)
```
