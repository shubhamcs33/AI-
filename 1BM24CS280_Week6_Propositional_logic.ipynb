```python

from itertools import product

# Implication operator
def implies(a, b):
    return (not a) or b


# Knowledge Base
def check_kb(P, Q, R):
    sentence1 = implies(Q, P)       # Q -> P
    sentence2 = implies(P, not Q)   # P -> NOT Q
    sentence3 = Q or R              # Q OR R

    KB = sentence1 and sentence2 and sentence3

    return sentence1, sentence2, sentence3, KB


# Truth Table Enumeration
def truth_table():
    print("P Q R | Q->P | P->~Q | QvR | KB | R | R->P | Q->R")
    print("-" * 60)

    models = []

    for P, Q, R in product([False, True], repeat=3):

        s1, s2, s3, KB = check_kb(P, Q, R)

        query1 = R
        query2 = implies(R, P)
        query3 = implies(Q, R)

        print(
            f"{'T' if P else 'F'} "
            f"{'T' if Q else 'F'} "
            f"{'T' if R else 'F'} |"
            f"  {'T' if s1 else 'F'}   |"
            f"   {'T' if s2 else 'F'}   |"
            f" {'T' if s3 else 'F'}  |"
            f" {'T' if KB else 'F'} |"
            f" {'T' if query1 else 'F'} |"
            f"  {'T' if query2 else 'F'}  |"
            f"  {'T' if query3 else 'F'}"
        )

        if KB:
            models.append((P, Q, R, query1, query2, query3))

    print("\nModels where KB is TRUE:")
    for model in models:
        print(model[:3])

    # Check entailment
    entails_R = all(model[3] for model in models)
    entails_R_P = all(model[4] for model in models)
    entails_Q_R = all(model[5] for model in models)

    print("\nEntailment Results:")
    print("KB entails R:", entails_R)
    print("KB entails R -> P:", entails_R_P)
    print("KB entails Q -> R:", entails_Q_R)


# Main program
truth_table()
```
