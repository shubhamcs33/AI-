```python

import random

N = 4

def calculate_cost(board):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if board[i] == board[j]:
                conflicts += 1
            elif abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def get_neighbors(board):
    neighbors = []

    for col in range(N):
        for row in range(N):
            if row != board[col]:
                new_board = board.copy()
                new_board[col] = row
                neighbors.append(new_board)

    return neighbors


def hill_climbing(board):
    while True:
        print("\nCurrent state:", board)
        print("Current cost:", calculate_cost(board))

        if calculate_cost(board) == 0:
            print("\nGoal state found!")
            print("Solution:", board)
            return

        neighbors = get_neighbors(board)
        best_cost = min(calculate_cost(n) for n in neighbors)

        best_neighbors = [
            n for n in neighbors
            if calculate_cost(n) == best_cost
        ]

        print("\nBest neighbor cost:", best_cost)
        for n in best_neighbors:
            print("Best neighbor:", n)

        if best_cost < calculate_cost(board):
            board = best_neighbors[0]
        else:
            print("Local minimum reached. Restarting...")
            board = [random.randint(0, N - 1)
                     for _ in range(N)]


print("Enter row positions for 4 columns (0 to 3):")
board = []

for col in range(N):
    row = int(input(f"Column {col + 1}: "))
    if row not in range(N):
        print("Invalid input. Enter 0, 1, 2, or 3.")
        exit()
    board.append(row)

hill_climbing(board)
```
